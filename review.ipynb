## Data Types
Python has several built-in data types. Knowing how they work is essential for writing correct code.
- **int**: whole numbers (e.g., `5`, `-2`)
- **float**: decimal numbers (e.g., `3.14`, `-0.5`)
- **str**: text (e.g., `'hello'`)
- **bool**: `True` or `False`
- **list**: ordered, mutable collection (e.g., `[1, 2, 3]`)
- **tuple**: ordered, immutable collection (e.g., `(1, 2, 3)`)
- **dict**: key-value pairs (e.g., `{'a': 1, 'b': 2}`)

## Control Flow: for, if, while
Control flow lets you decide what your code does and how often it runs.
- **for loops**: repeat for each item in a sequence (list, string, etc.)
    ```python
    for i in range(3):
        print(i)  # prints 0, 1, 2
    ```
- **if statements**: run code only if a condition is true
    ```python
    if x > 0:
        print('Positive!')
    ```
- **while loops**: repeat as long as a condition is true
    ```python
    count = 0
    while count < 3:
        print(count)
        count += 1
    ```
- Indentation matters!

Use a `for` loop when iterating through a sequence or a known range. Use a `while` loop when repetition should continue until a condition changes.

````python
# Write a for loop that prints the even numbers up to 20

for i in range(0,21,2):
    print(i)



# Write a while loop that counts down from 10 to 0.
i = 10
while i>=0:
    print(i)
    i-=1

````

### The `break` and `continue` Statements in Loops
- `break` immediately exits the loop, skipping all remaining iterations.
- `continue` skips the rest of the code in the current iteration and moves to the next one.
Use these to control the flow of your loops!

````python
# Example: Print only odd numbers from 1 to 10 using continue
for i in range(11):
    if i%2==0:
        continue
    print(i)
````

````python
# increment and print a counter repeatedly
# starting from 0, with a 20% random chance
# of stopping the counter and ending the code.

import random as random

counter = 0

while True:
    print(counter)
    if random.random() < 0.2:
        break
    counter += 1
````

## Numpy Arrays: Initializing, Indexing, Slicing
Numpy arrays are powerful for numerical data. They are like lists, but faster and support math operations directly.
- **Import numpy**: `import numpy as np`
- **Create arrays**:
    - `np.array([1, 2, 3])`
    - `np.zeros(4)` (array of zeros)
    - `np.ones((2, 3))` (2x3 array of ones)
    - `np.arange(5)` (array: 0,1,2,3,4)
- **Indexing**: starts at 0. `A[1]` is the second element.
- **Slicing**: `A[1:4]` gives elements 1,2,3 (not 4).
- **Arrays are mutable**: changing one element changes the array.
- **Broadcasting**: you can add, subtract, multiply arrays and numbers directly.
**Tips:**
- Use `.shape` to check array dimensions.
- Use `.copy()` to avoid reference issues.

````python
# Import NumPy and create a 1D array with the integers from 1 through 20.
import numpy as np

arr = np.arange(1,21,dtype= np.int32)
print(arr)

# Print the first five elements and the final three elements of the array.
print(arr[:5])
print(arr[-3:])

# Use a Boolean condition to print the values that are both even and greater than 10.
bool_mask = (arr[:] > 10) & (arr[:]%2==0)
print(bool_mask)
print(arr[bool_mask])

# Change the first element to 100, reshape the array into a 4-by-5 array, and print the result.
arr[0] = 100
print(arr)
arr = arr.reshape(4,5)
print(arr.shape)
````

## Reference Behavior & Mutability
In Python, some objects are **mutable** (can change), and some are **immutable** (cannot change). Lists, numpy arrays, and pandas DataFrames are mutable.
- **Assignment does NOT copy!** `B = A` means B and A point to the same object.
- To make a true copy:
    - For lists, arrays, or df's: `B = A.copy()`
- Changing one will change the other if you didn't copy!
Example:
```python
A = [1, 2, 3]
B = A
B[0] = 99
print(A)  # [99, 2, 3]
```
**Tips:**
- Use `id(x)` to check if two variables point to the same object.
- Use `.copy()` to avoid accidental changes.
